```python
from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_DIR = Path.cwd().parent

RT_SAMPLE_DIR = (
    PROJECT_DIR
    / "data"
    / "realtime"
    / "sample_2026-11-09"
)

files = {
    "raw_stop_updates": RT_SAMPLE_DIR / "rt_stop_updates_raw.csv",
    "final_stop_events": RT_SAMPLE_DIR / "rt_stop_events_final.csv",
    "trip_day": RT_SAMPLE_DIR / "rt_trip_day.csv"
}

for name, path in files.items():

    print("\n" + "=" * 70)
    print(name.upper())
    print("=" * 70)

    # Inspect column names without loading the full file
    columns = pd.read_csv(path, nrows=0).columns.tolist()

    print("\nColumns:")
    print(columns)

    # Load only five sample rows
    sample = pd.read_csv(path, nrows=5)

    print("\nFirst five rows:")
    display(sample)

    print("\nData types in sample:")
    print(sample.dtypes)
```

```python
# Preserve identifiers as strings
common_dtypes = {
    "trip_id": "string",
    "start_date": "string",
    "trip_status": "string"
}

raw_dtypes = {
    **common_dtypes,
    "stop_id": "string",
    "stop_sequence": "Int64",
    "stop_status": "string"
}

final_dtypes = {
    **common_dtypes,
    "stop_sequence": "Int64",
    "stop_status": "string"
}

# Load raw stop updates
rt_raw = pd.read_csv(
    files["raw_stop_updates"],
    dtype=raw_dtypes,
    low_memory=False
)

# Load final stop events
rt_final = pd.read_csv(
    files["final_stop_events"],
    dtype=final_dtypes,
    low_memory=False
)

# Load trip-day data
rt_trip_day = pd.read_csv(
    files["trip_day"],
    dtype=common_dtypes,
    low_memory=False
)

# Parse timestamps
raw_datetime_cols = [
    "snapshot_time",
    "predicted_arrival",
    "predicted_departure"
]

final_datetime_cols = [
    "planned_arrival",
    "planned_departure",
    "first_seen",
    "last_seen"
]

trip_datetime_cols = ["first_seen", "last_seen"]

for col in raw_datetime_cols:
    rt_raw[col] = pd.to_datetime(rt_raw[col], errors="coerce")

for col in final_datetime_cols:
    rt_final[col] = pd.to_datetime(rt_final[col], errors="coerce")

for col in trip_datetime_cols:
    rt_trip_day[col] = pd.to_datetime(
        rt_trip_day[col], errors="coerce"
    )

print("Raw stop updates:", rt_raw.shape)
print("Final stop events:", rt_final.shape)
print("Trip-day:", rt_trip_day.shape)
```

```python
datasets = {
    "RAW STOP UPDATES": rt_raw,
    "FINAL STOP EVENTS": rt_final,
    "TRIP DAY": rt_trip_day
}

for name, df in datasets.items():

    print("\n" + "=" * 65)
    print(name)
    print("=" * 65)

    print("Rows:", len(df))
    print("Columns:", len(df.columns))

    print("\nMissing values:")
    display(
        pd.DataFrame({
            "missing_count": df.isna().sum(),
            "missing_percentage": (
                df.isna().mean() * 100
            ).round(2)
        }).sort_values(
            "missing_percentage",
            ascending=False
        )
    )

    print("\nDuplicate complete rows:", df.duplicated().sum())
```

```python
# Candidate business keys
key_checks = {
    "RAW STOP UPDATES": (
        rt_raw,
        ["snapshot_time", "trip_id", "start_date", "stop_sequence"]
    ),

    "FINAL STOP EVENTS": (
        rt_final,
        ["trip_id", "start_date", "stop_sequence"]
    ),

    "TRIP DAY": (
        rt_trip_day,
        ["trip_id", "start_date"]
    )
}

for name, (df, key_cols) in key_checks.items():

    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    duplicate_mask = df.duplicated(
        subset=key_cols,
        keep=False
    )

    print("Rows:", len(df))
    print("Unique key combinations:", df[key_cols].drop_duplicates().shape[0])
    print("Rows involved in duplicate keys:", duplicate_mask.sum())

    if duplicate_mask.any():
        display(
            df.loc[duplicate_mask, key_cols]
            .sort_values(key_cols)
            .head(10)
        )
```

```python
print("Unique snapshots:", rt_raw["snapshot_time"].nunique())

print("First snapshot:", rt_raw["snapshot_time"].min())
print("Last snapshot:", rt_raw["snapshot_time"].max())

print("\nTrip statuses in raw data:")
display(rt_raw["trip_status"].value_counts(dropna=False))

print("\nStop statuses in raw data:")
display(rt_raw["stop_status"].value_counts(dropna=False))

print("\nTrip statuses in final events:")
display(rt_final["trip_status"].value_counts(dropna=False))

print("\nStop statuses in final events:")
display(rt_final["stop_status"].value_counts(dropna=False))
```

```python
# Ensure delay columns are numeric
delay_cols = [
    "arrival_delay_s",
    "departure_delay_s"
]

for col in delay_cols:
    rt_raw[col] = pd.to_numeric(
        rt_raw[col],
        errors="coerce"
    )

    rt_final[col] = pd.to_numeric(
        rt_final[col],
        errors="coerce"
    )

# Analyse ordinary scheduled stop observations
raw_scheduled = rt_raw[
    rt_raw["stop_status"] == "SCHEDULED"
].copy()

final_scheduled = rt_final[
    rt_final["stop_status"] == "SCHEDULED"
].copy()

print("Raw scheduled observations:", len(raw_scheduled))
print("Final scheduled events:", len(final_scheduled))
```

```python
def delay_summary(df, dataset_name):

    results = []

    for col in ["arrival_delay_s", "departure_delay_s"]:

        values = df[col].dropna()

        results.append({
            "dataset": dataset_name,
            "delay_type": col,
            "available_values": len(values),
            "missing_values": df[col].isna().sum(),
            "mean_seconds": values.mean(),
            "median_seconds": values.median(),
            "p90_seconds": values.quantile(0.90),
            "p95_seconds": values.quantile(0.95),
            "minimum_seconds": values.min(),
            "maximum_seconds": values.max()
        })

    return pd.DataFrame(results)


delay_stats = pd.concat(
    [
        delay_summary(raw_scheduled, "Raw"),
        delay_summary(final_scheduled, "Final")
    ],
    ignore_index=True
)

display(delay_stats.round(2))
```

```python
def delay_categories(df, col):

    values = df[col].dropna()

    return pd.Series({
        "Early": (values < 0).sum(),
        "On time": (values == 0).sum(),
        "Delayed": (values > 0).sum(),
        "Missing": df[col].isna().sum()
    })


for col in ["arrival_delay_s", "departure_delay_s"]:

    print("\n", col)

    comparison = pd.DataFrame({
        "Raw": delay_categories(raw_scheduled, col),
        "Final": delay_categories(final_scheduled, col)
    })

    display(comparison)
```

```python
import pandas as pd
from pathlib import Path

PROJECT_DIR = Path.cwd().parent
GTFS_DIR = PROJECT_DIR / "extracted"

stops = pd.read_csv(
    GTFS_DIR / "stops.txt",
    dtype={
        "stop_id": "string",
        "parent_station": "string",
        "location_type": "Int64",
        "platform_code": "string"
    }
)

stop_times = pd.read_csv(
    GTFS_DIR / "stop_times.txt",
    dtype={
        "trip_id": "string",
        "stop_id": "string",
        "arrival_time": "string",
        "departure_time": "string",
        "stop_sequence": "Int64"
    }
)

trips = pd.read_csv(
    GTFS_DIR / "trips.txt",
    dtype={
        "trip_id": "string",
        "route_id": "string",
        "service_id": "string"
    }
)

routes = pd.read_csv(
    GTFS_DIR / "routes.txt",
    dtype={
        "route_id": "string",
        "agency_id": "string",
        "route_short_name": "string"
    }
)

print("Static GTFS files loaded.")
```

```python
scheduled = (
    stop_times
    .merge(
        trips,
        on="trip_id",
        how="left",
        validate="many_to_one"
    )
    .merge(
        routes,
        on="route_id",
        how="left",
        validate="many_to_one"
    )
    .merge(
        stops[["stop_id", "stop_name", "parent_station"]],
        on="stop_id",
        how="left",
        validate="many_to_one"
    )
)

print("Scheduled timetable:", scheduled.shape)
```

```python
other_trip_ids = ["784329", "1572674", "418689"]

other_trips = rt_raw[
    rt_raw["trip_id"].isin(other_trip_ids)
].copy()

# Select the earliest snapshot for each trip
first_snapshot = (
    other_trips.groupby(["trip_id", "start_date"])["snapshot_time"]
    .min()
    .reset_index()
)

other_sample = other_trips.merge(
    first_snapshot,
    on=["trip_id", "start_date", "snapshot_time"],
    how="inner"
)

display(
    other_sample[
        [
            "trip_id",
            "start_date",
            "snapshot_time",
            "stop_sequence",
            "stop_id",
            "arrival_delay_s",
            "departure_delay_s",
            "predicted_arrival",
            "predicted_departure"
        ]
    ].sort_values(["trip_id", "stop_sequence"])
)
```

```python
# Keep all four trips with extreme arrival-delay observations
affected_trip_ids = ["497812", "784329", "1572674", "418689"]

affected = rt_raw[
    (rt_raw["trip_id"].isin(affected_trip_ids)) &
    (rt_raw["arrival_delay_s"] <= -3600)
].copy()

# Keep the earliest observation for each trip-date-stop combination
affected_first = (
    affected.sort_values("snapshot_time")
    .drop_duplicates(
        subset=["trip_id", "start_date", "stop_sequence", "stop_id"],
        keep="first"
    )
)

# Retrieve the corresponding static timetable
schedule_lookup = stop_times[
    [
        "trip_id",
        "stop_sequence",
        "stop_id",
        "arrival_time",
        "departure_time"
    ]
].copy()

comparison = affected_first.merge(
    schedule_lookup,
    on=["trip_id", "stop_sequence", "stop_id"],
    how="left",
    validate="one_to_one"
)

# Add stop names
comparison = comparison.merge(
    stops[["stop_id", "stop_name"]],
    on="stop_id",
    how="left",
    validate="many_to_one"
)

display(
    comparison[
        [
            "trip_id",
            "start_date",
            "stop_sequence",
            "stop_name",
            "arrival_time",
            "departure_time",
            "arrival_delay_s",
            "predicted_arrival",
            "snapshot_time"
        ]
    ].sort_values(["trip_id", "stop_sequence"])
)
```

```python
# Convert GTFS service times into actual scheduled datetimes.
# For example, 24:08:00 becomes midnight + 8 minutes on the next day.

comparison["scheduled_arrival_dt"] = (
    pd.to_datetime(comparison["start_date"])
    + pd.to_timedelta(comparison["arrival_time"])
)

# Calculate predicted minus scheduled time.
# This is a provisional comparison because timezone handling
# still needs to be validated.
comparison["calculated_delay_s"] = (
    comparison["predicted_arrival"]
    - comparison["scheduled_arrival_dt"]
).dt.total_seconds()

comparison["delay_difference_s"] = (
    comparison["arrival_delay_s"]
    - comparison["calculated_delay_s"]
)

display(
    comparison[
        [
            "trip_id",
            "stop_sequence",
            "stop_name",
            "scheduled_arrival_dt",
            "predicted_arrival",
            "arrival_delay_s",
            "calculated_delay_s",
            "delay_difference_s"
        ]
    ].sort_values(["trip_id", "stop_sequence"])
)
```

```python
# Create a timezone-aware scheduled arrival in German local time
comparison["scheduled_arrival_local"] = (
    pd.to_datetime(comparison["start_date"])
    .dt.tz_localize("Europe/Berlin")
    + pd.to_timedelta(comparison["arrival_time"])
)

# Interpret the original CSV timestamps as UTC
comparison["predicted_arrival_utc"] = pd.to_datetime(
    comparison["predicted_arrival"],
    utc=True
)

# Convert predictions to German local time
comparison["predicted_arrival_local"] = (
    comparison["predicted_arrival_utc"]
    .dt.tz_convert("Europe/Berlin")
)

# Recalculate delay using timezone-aware timestamps
comparison["calculated_delay_local_s"] = (
    comparison["predicted_arrival_local"]
    - comparison["scheduled_arrival_local"]
).dt.total_seconds()

# Compare with provider-reported delay
comparison["delay_difference_local_s"] = (
    comparison["arrival_delay_s"]
    - comparison["calculated_delay_local_s"]
)

display(
    comparison[
        [
            "trip_id",
            "stop_sequence",
            "stop_name",
            "scheduled_arrival_local",
            "predicted_arrival_local",
            "arrival_delay_s",
            "calculated_delay_local_s",
            "delay_difference_local_s"
        ]
    ].sort_values(["trip_id", "stop_sequence"])
)
```
